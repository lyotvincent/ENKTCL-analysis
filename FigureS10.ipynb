### Quality control (QC) metrics for NK and T cell subclusters (Figure S10)

(a)-(f) Violin plots display the distribution of (a, d) log-transformed total UMI counts (log nUMI), (b, e) log10-transformed number of detected genes (log nGenes), and (c, f) mitochondrial gene fraction (MT fraction) across NK cell (a–c) and T cell (d–f) subclusters, respectively.

```python
import anndata as ad
import matplotlib.pyplot as plt
import pandas as pd
import scanpy as sc
import numpy as np
import matplotlib as mpl
import matplotlib.ticker as ticker
import sys
import os
mpl.rcParams['font.family'] = 'Arial'
plt.rcParams["font.sans-serif"] = ["Arial"]
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["font.size"] = 14
print(sys.version)
```

```python
figPath = "./figures/figS10"
NKPath = './data/NKTCL_NK.h5ad'
TPath = './data/NKTCL_T.h5ad'
os.makedirs(figPath, exist_ok=True)
```

```python
adata_NK = sc.read_h5ad(NKPath)
adata_T = sc.read_h5ad(TPath)
```

#### **Fig. S10a** Violin plots display the distribution of log-transformed total UMI counts (nUMIs) in NK cell states

```python
for gene in ['log1p_total_counts']:
  fig, ax = plt.subplots(1, 1, figsize=(4, 4))
  sc.pl.violin(adata_NK, groupby='CellStateLabels', keys=gene, ax=ax, show=False,palette=['#66c2a5', '#fc8d62', 
          '#8da0cb', '#e78ac3', '#8fb74d', '#a6d854','#ffd92f', '#e5c494', '#b3b3b3'])
  ax.spines['right'].set_visible(False)
  ax.spines['top'].set_visible(False)
  ax.set_xlabel("")
  ax.set_ylabel(gene)
  ax.set_xticklabels(adata_NK.obs['CellStateLabels'].cat.categories, rotation=90)
  fig.savefig(f"{figPath}/FigS10a.png", dpi=150, bbox_inches='tight')
```

#### **Fig. S10b** Violin plots display the distribution of log-transformed number of detected genes (nGenes) in NK cell states

```python
for gene in ['log1p_n_genes_by_counts']:
  fig, ax = plt.subplots(1, 1, figsize=(4, 4))
  sc.pl.violin(adata_NK, groupby='CellStateLabels', keys=gene, ax=ax, show=False,palette=['#66c2a5', '#fc8d62', 
          '#8da0cb', '#e78ac3', '#8fb74d', '#a6d854','#ffd92f', '#e5c494', '#b3b3b3'])
  ax.spines['right'].set_visible(False)
  ax.spines['top'].set_visible(False)
  ax.set_xlabel("")
  ax.set_ylabel(gene)
  ax.set_xticklabels(adata_NK.obs['CellStateLabels'].cat.categories, rotation=90)
  # ax.legend(  handletextpad=0.5, frameon=False,
  #                     borderpad=0.6,
  #                     columnspacing=1.3,
  #                     handlelength=0.65, fontsize=12)
  fig.savefig(f"{figPath}/FigS10b.png", dpi=150, bbox_inches='tight')
```
